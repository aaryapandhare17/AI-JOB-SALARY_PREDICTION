```python
def extract_job(job, search_role):
    clean_job = {
        "job_id": job.get("id"),
        "search_role": search_role,
        "title": job.get("title"),
        "company": job.get("company", {}).get("display_name"),
        "location": job.get("location", {}).get("display_name"),
        "category": job.get("category", {}).get("label"),
        "contract_time": job.get("contract_time"),
        "description": job.get("description"),
        "salary_min": job.get("salary_min"),
        "salary_max": job.get("salary_max"),
        "salary_is_predicted": job.get("salary_is_predicted"),
        "created": job.get("created"),
        "redirect_url": job.get("redirect_url")
    }

    return clean_job
```

```python
roles = [
    "software engineer",
    "data analyst",
    "data scientist",
    "machine learning engineer",
    "ai engineer",
    "data engineer",
    "backend developer",
    "frontend developer",
    "full stack developer",
    "devops engineer",
    "cloud engineer",
    "cybersecurity analyst"
]

print("Number of roles:", len(roles))
```

```python
def fetch_jobs(role, page=1, results_per_page=50):

    url = f"https://api.adzuna.com/v1/api/jobs/in/search/{page}"

    params = {
        "app_id": APP_ID,
        "app_key": APP_KEY,
        "results_per_page": results_per_page,
        "what": role,
        "salary_min": 1
    }

    response = requests.get(url, params=params, timeout=30)

    if response.status_code == 200:
        data = response.json()
        return data.get("results", [])

    print(
        f"Request failed for {role}, "
        f"page {page}: {response.status_code}"
    )

    return []
```

```python
all_jobs = []

pages_per_role = 5

for role in roles:

    print(f"\nCollecting: {role}")

    for page in range(1, pages_per_role + 1):

        jobs = fetch_jobs(role, page)

        print(f"  Page {page}: {len(jobs)} jobs")

        for job in jobs:

            if (
                job.get("salary_min") is not None
                and job.get("salary_max") is not None
                and job.get("salary_is_predicted") == "0"
            ):
                clean_job = extract_job(job, role)
                all_jobs.append(clean_job)
```

```python
jobs_df = pd.DataFrame(all_jobs)

print("Dataset shape:", jobs_df.shape)

jobs_df.head()
```
